```python
import pandas as pd
from xgboost import XGBClassifier,XGBRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
```

```python
# pip install xgboost
```

```python
df=pd.read_csv('loan_approval_dataset.csv')
```

```python
df.info()
```

```python
df.head()
```

```python
x=df.drop(columns='Loan_Approved')
```

```python
df['Loan_Approved']=df['Loan_Approved'].map({'Yes':1,'No':0})
```

```python
y=df['Loan_Approved']
```

```python
obj_cols=x.select_dtypes(include='object').columns
```

```python
x[obj_cols]=x[obj_cols].astype(dtype='category')
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
model=XGBClassifier(enable_categorical=True)
```

```python
model.fit(x_train,y_train)
```

```python
pred_ytrain=model.predict(x_train)
```

```python
pred_ytest=model.predict(x_test)
```

```python
print(classification_report(pred_ytrain,pred_ytrain))
```

```python
print(classification_report(pred_ytest,pred_ytest))
```
